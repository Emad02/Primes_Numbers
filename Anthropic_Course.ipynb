```python
# Creat an AIP Client
import anthropic

client = anthropic.Anthropic(api_key = api_key)
model = "claude-sonnet-4-6"
```

```python
# Make a request

message = client.messages.create(
    model = model,
    max_tokens= 5000,
    messages=[
        {"role": "user",
        "content": "recently, I read an article that was in the field of Chemistry. On Data exploration section, researcher used PCA and UMAP embedding to understand the clusters and groups. Could you tell me that I can use PHATE method instead of the used approaches? If yes, provide a behind steps of this process in ten lines. In addition, think realistic."}
    ]
)
```

```python
# Make a chat

def add_user_message(messages, text):
  user_message = {"role": "user", "content": text}
  messages.append(user_message)

def add_assistant_message(messages, text):
  assistant_message = {"role": "assistant", "content": text}
  messages.append(assistant_message)


def chat(messages):
  message = client.messages.create(
      model = model,
      max_tokens= 5000,
      messages=messages,
  )
  return message.content[0].text
```

```python
# Make a starting list of messages
messages = []

# Add in the initial user question of "recently, I read an article that was in the field of Chemistry. On Data exploration section, researcher used PCA and UMAP embedding to understand the clusters and groups. Could you tell me that I can use PHATE method instead of the used approaches? If yes, provide a behind steps of this process in ten lines. In addition, think realistic."
add_user_message(messages, "recently, I read an article that was in the field of Chemistry. On Data exploration section, researcher used PCA and UMAP embedding to understand the clusters and groups. Could you tell me that I can use PHATE method instead of the used approaches? If yes, provide a behind steps of this process in ten lines. In addition, think realistic.")

# Pass the list of the messages into 'chat' to get an answer
answer = chat(messages)

# Take the answer and add it as an assistant message into our list
add_assistant_message(messages, answer)

# Add in the user's follow-up question
add_user_message(messages, "Convert your response to a latex code to run it on Overleaf")

# Call chat again with the list of messages to get a final answer
answer = chat(messages)
answer
```
